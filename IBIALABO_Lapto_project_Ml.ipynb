## Feature Engineering

```python
data.info()
```

```python
data.ScreenResolution.unique()
```

```python
data["screen_w&h"]=data["ScreenResolution"].str.extract(r'(\d+x\d+)')
data["IPS"]=data["ScreenResolution"].apply(lambda x:1 if 'IPS'in x else 0)
data["Touchscreen"]=data["ScreenResolution"].apply(lambda x:1 if 'Touchscreen'in x else 0)
```

```python
data.head(3)
```

```python
data["width"]=data["screen_w&h"].str.split("x").str[0].astype(int)
data["height"]=data["screen_w&h"].str.split("x").str[1].astype(int)
```

```python
data.head(2)
```

```python
data.drop(columns=["ScreenResolution","screen_w&h"],inplace=True)
```

```python
data.Cpu.unique()
```

```python
data["cpu_brands"]=data["Cpu"].apply(lambda x:" ".join(x.split()[0:3]))
```

```python
data.head(3)
```

```python
def cpu_brand_name(text):
    if text=="Intel Core i5" or text=="Intel Core i3" or text=="Intel Core i7":
        return text
    elif text.startswith("AMD"):
        return "AMD"
    elif text.startswith("Samsung"):
        return "Samsung"
    else:
        return "other intel"
        
    
```

```python
data["cpu_brand_name"]=data["cpu_brands"].apply(cpu_brand_name)
```

```python
data["speed"]=data["Cpu"].str.split().str[-1]
```

```python
data["speed"]=data["speed"].str.replace("GHz","").astype(float)
```

```python
data.speed.isna().sum()
```

```python
data.head(3)
```

```python
data.drop(columns=["Cpu","cpu_brands"],inplace=True)
```

```python
data.Ram.unique()
```

```python
data.Ram=data["Ram"].str.replace("GB","").astype(int)
```

```python
data.head(2)
```

```python
data.Memory.unique()
```

```python
#replacing my .0 with nothing,replacing TB with 024 because a standard company strict coversion
#of 1TB=1024 but its usually 1000 and finally replacing my GB with nothing that means converting all to TB

data.Memory=data.Memory.str.replace(".0","")
data.Memory=data.Memory.str.replace("TB","024")
data.Memory=data.Memory.str.replace("GB","")
```

```python
data.Memory.unique()
```

```python
store=data["Memory"].str.split("+",n=1,expand=True)
data["first_memory"]=store[0]
data["second_memory"]=store[1]
```

```python
data.fillna({"second_memory":"0"},inplace=True)
```

```python
data['SSD_FIRST_MEMEROY']=data["first_memory"].apply(lambda x:1 if "SSD" in x else 0)
data['HDD_FIRST_MEMORY']=data["first_memory"].apply(lambda x:1 if "HDD" in x else 0)
data['FLASH_STORAGE_FIRST_MEMORY']=data["first_memory"].apply(lambda x:1 if "Flash Storage" in x else 0)
data['Hybrid_FIRST_MEMORY']=data["first_memory"].apply(lambda x:1 if "Hybrid" in x else 0)
```

```python
data['SSD_SECOND_MEMEROY']=data["second_memory"].apply(lambda x:1 if "SSD" in x else 0)
data['HDD_SECOND_MEMORY']=data["second_memory"].apply(lambda x:1 if "HDD" in x else 0)
data['FLASH_STORAGE_SECOND_MEMORY']=data["second_memory"].apply(lambda x:1 if "Flash Storage" in x else 0)
data['Hybrid_SECOND_MEMORY']=data["second_memory"].apply(lambda x:1 if "Hybrid" in x else 0)
```

```python
data["second_memory"].unique()
```

```python
data["first_memory"]=data["first_memory"].str.extract(r'(\d+)').astype(int)

                                                   
```

```python
data["second_memory"]=data["second_memory"].str.extract(r'(\d+)').astype(int)
```

```python
data["SSD"]=data['first_memory']*data['SSD_FIRST_MEMEROY']+data['second_memory']*data['SSD_SECOND_MEMEROY'] 
data["HDD"]=data['first_memory']*data['HDD_FIRST_MEMORY']+data['second_memory']*data['HDD_SECOND_MEMORY']
data["flash_storage"]=data['first_memory']*data['FLASH_STORAGE_FIRST_MEMORY']+data['second_memory']*data['FLASH_STORAGE_SECOND_MEMORY']
data["Hybrid"]=data['first_memory']*data['Hybrid_FIRST_MEMORY']+data['second_memory']*data['Hybrid_SECOND_MEMORY']    
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error,r2_score, mean_absolute_error
```

```python
lr_pipeline=Pipeline(steps=[
        ("model",LinearRegression())])

dr_pipeline=Pipeline(steps=[
        ("model",DecisionTreeRegressor(random_state=42))])

rf_pipeline=Pipeline(steps=[
        ("model",RandomForestRegressor(random_state=42))])
```

## Using Label Encoding 

```python
label_encode=data.copy()
```

```python
from sklearn.preprocessing import LabelEncoder
def label_encoder(datas):
    datas=datas.copy()
    le=LabelEncoder()
    categorical_col=datas.select_dtypes(include="object").columns
    for cols in categorical_col:
        datas[cols]=le.fit_transform(datas[cols])
    return datas
```

```python
le_encoded=label_encoder(label_encode)
le_encoded.head(2)
```

```python
y_la=le_encoded["Price"]
```

```python
x_la=le_encoded.drop(columns="Price")
```

```python
from sklearn.model_selection import train_test_split
xla_train,xla_test,yla_train,yla_test=train_test_split(x_la,y_la,test_size=0.2,random_state=42)
```

```python
result_la={}
```

```python
models={"Linear Regression ":LinearRegression(),
        "Decision Tree ":DecisionTreeRegressor(random_state=42),
        "Random Forest ":RandomForestRegressor(random_state=42),}
```

```python
for name,model in models.items():
    pipe=Pipeline(steps=[
        ("model",model)])
    pipe.fit(xla_train,yla_train)
    yla_pred=pipe.predict(xla_test)

    result_la[name]={"R2":r2_score(yla_test,yla_pred),
                  "MAE":mean_absolute_error(yla_test,yla_pred)}

```

```python
result_df=pd.DataFrame(result_la).T
print(result_df)
```
